```python
index = 0
s = "BlockDMask"
for a in s:
    if a == 'k':
        break    # 'k'를 찾았으니 for문에서 나와라!

    index = index + 1


print(index)    # 'k'가 첫번째로 존재하는 위치 출력
```

```python
result = 0
for a in range(1,101):  # 1 ~ 100
    result = result + a    # a를 더해주고

    if result > 100:    # result가 100이 넘었을때
        print(a)        # 그때의 a값을 출력
        break

print(result)
```

```python
sr = ['father', 'mother', 'brother'] #r을 추가하면 cnt의 개수도 늘어남
cnt = 0
for s in sr:
        for c in s:
            if c == 'r': #c가 r이라는 알파벳일 때
                cnt += 1 #cnt에 1 추가
print(cnt)
```

```python
a = []    # 빈 리스트 생성

for i in range(3):
    line = []              # 안쪽 리스트로 사용할 빈 리스트 생성
    for j in range(2):
        line.append(0)     # 안쪽 리스트에 0 추가
    a.append(line)         # 전체 리스트에 안쪽 리스트를 추가

print(a)
```
